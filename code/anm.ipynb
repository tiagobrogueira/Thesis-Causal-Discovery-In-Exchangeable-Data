```python
import numpy as np
import nbimporter
import os
import random
from sklearn.linear_model import LinearRegression
import cepairsimplementation as ce
seedR = random.Random(42)
seedN = np.random.default_rng()
from sklearn.gaussian_process import GaussianProcessRegressor
```

```python
def anm_score(x,y,test,model):
    if x.shape[1]>1 or y.shape[1]>1:
        return np.nan
    x = (x - np.mean(x)) / np.std(x)
    y = (y - np.mean(y)) / np.std(y)
    if model=="gp":
        gp = GaussianProcessRegressor().fit(x, y)
        y_predict = gp.predict(x)
    else:
        reg=LinearRegression().fit(x,y)
        y_predict = reg.predict(x)
    errors=(y - y_predict.reshape(-1,1))**2
    result=ce.test_independence(x,errors, method=test)
    return result


def anm(d,test="hoef",model="l"):
    x,y=d
    return anm_score(x,y,test,model)-anm_score(y,x,test,model)
```
